1. Read 5 numbers and print all array elements.


```python
def read_array(n):
    arr = []
    for i in range(0,n):
        element = int(input(f"Enter array element {i+1}: ;"))
        arr.append(element)
    return arr

def printAllArray(arr,n):
    for i in range(0,n):
        print(arr[i])

n = int(input("Enter size of array: "))
arr = read_array(n)
printAllArray(arr,n)
    
```

2. Read 5 numbers and find their sum.

```python
def sumOfArray(arr,n):
    total = 0
    for i in range(0,n):
        total += arr[i]
    return total

n = int(input("Enter size of array: "))
arr = read_array(n)
print(sumOfArray(arr,n))
```

3. Find the average of array elements.

```python
def AverageOfArray(arr,n):
    Average = sumOfArray(arr,n)//n
    return Average

n = int(input("Enter size of array: "))
arr = read_array(n)
print(AverageOfArray(arr,n))
```

4. Find the largest number in an array.

```python
def LargestOfArray(arr,n):
    largest = arr[0]
    for i in range(1,n):
        if (arr[i] > largest):
            largest = arr[i]
    return largest

n = int(input("Enter size of array: "))
arr = read_array(n)
print("Largest No Of Array is: ", LargestOfArray(arr,n))
```

####
5. Find the smallest number in an array.

```python
def SmallestOfArray(arr,n):
    Smallest = arr[0]
    for i in range(1,n):
        if (arr[i] < Smallest):
            Smallest = arr[i]
    return Smallest

n = int(input("Enter size of array: "))
arr = read_array(n)
print("Smallest No Of Array is: ", SmallestOfArray(arr,n))
```
